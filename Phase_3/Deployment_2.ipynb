```python
import pandas as pd
import joblib


model = joblib.load("advanced_water_model.pkl")

scaler = joblib.load("advanced_scaler.pkl")


household = int(input("Enter Household Size: "))

season = float(input("Enter Seasonal Index: "))

garden = float(input("Enter Garden Area: "))

income = float(input("Enter Income Level: "))

bathrooms = int(input("Enter Number of Bathrooms: "))

appliances = int(input("Enter Appliances Count: "))

rainfall = float(input("Enter Rainfall: "))

temperature = float(input("Enter Temperature: "))

price = float(input("Enter Water Price Index: "))


input_data = pd.DataFrame(
    [[
        household,
        season,
        garden,
        income,
        bathrooms,
        appliances,
        rainfall,
        temperature,
        price
    ]],

    columns=[
        "Household_Size",
        "Seasonal_Index",
        "Garden_Area",
        "Income_Level",
        "Bathrooms",
        "Appliances_Count",
        "Rainfall",
        "Temperature",
        "Water_Price_Index"
    ]
)


scaled_data = scaler.transform(input_data)


prediction = model.predict(scaled_data)

print(
    "\nPredicted Daily Water Usage:",
    round(prediction[0], 2),
    "liters"
)
```
